### Example 2: Variables and Basic Data Types

A **variable** is a name that stores a value, created with `=` (assignment). Unlike some other languages, you never have to declare a type in Python — it figures out the type automatically from the value you assign. This is called **dynamic typing**. The built-in function `type()` tells you the type of a value.

````python
# Some basic data types in Python
name = "Max"        # str -> text, always in quotes
age = 25             # int -> whole number
height = 1.68         # float -> number with a decimal point
is_student = True    # bool -> True or False only

# f-Strings (f"...") let us insert variables directly into a string using {}
print(f"Name: {name} (Type: {type(name)})")
print(f"Age: {age} (Type: {type(age)})")
print(f"Height: {height} (Type: {type(height)})")
print(f"Is Student: {is_student} (Type: {type(is_student)})")
````

Note: `list` and `dict` are also common data types in Python — we'll take a close look at them a bit later, in their own dedicated examples.

### Example 3: Basic Arithmetic Operators

Python supports the usual math operators. A few are less obvious: `//` rounds division down to a whole number ("floor division"), `%` gives you the remainder of a division ("modulo"), and `**` is used for exponents (powers).

````python
a = 17
b = 5

print(f"{a} + {b} = {a + b}")   # Addition
print(f"{a} - {b} = {a - b}")   # Subtraction
print(f"{a} * {b} = {a * b}")   # Multiplication
print(f"{a} / {b} = {a / b}")   # Division -> always returns a float
print(f"{a} // {b} = {a // b}") # Floor division -> rounds down to the nearest whole number
print(f"{a} % {b} = {a % b}")   # Modulo -> remainder of the division (17 = 3*5 + 2)
print(f"{a} ** 2 = {a ** 2}")   # Exponent -> a to the power of 2
````

### Example 4: Comparison and Logical Operators

Comparison operators (`==`, `!=`, `<`, `>`, `<=`, `>=`) compare two values and always give back a `bool` (`True` or `False`). Logical operators (`and`, `or`, `not`) combine several such `True`/`False` values. We need exactly these booleans in the next example to let our program make decisions.

````python
age = 20
has_ticket = True

# Comparison operators -> result is always True or False
print(f"age == 20: {age == 20}")   # equal to
print(f"age != 20: {age != 20}")   # not equal to
print(f"age >= 18: {age >= 18}")   # greater than or equal to

# Logical operators -> combine several True/False values
print(f"age >= 18 and has_ticket: {age >= 18 and has_ticket}") # both must be True
print(f"age < 18 or has_ticket: {age < 18 or has_ticket}")     # at least one must be True
print(f"not has_ticket: {not has_ticket}")                     # flips True to False and vice versa
````

### Example 5: Converting Between Data Types

Sometimes a value has the "wrong" type for what you want to do with it — for example, text read from a file is always a `str`, even if it looks like a number. `int()`, `float()`, and `str()` convert a value from one type into another.

````python
# Text that looks like a number is still just text (str) until we convert it
text_number = "5"
print(f"{text_number} has type {type(text_number)}")

converted_number = int(text_number)  # str -> int
print(f"{converted_number} has type {type(converted_number)}")

# A few more conversions
print(int("42") + 8)      # str -> int, then we can do math with it
print(float("3.14") * 2)  # str -> float
print(str(99) + "%")      # int -> str, so we can glue it to another string with +
print(int(7.9))           # float -> int simply cuts off the decimal part (no rounding!)

# Careful: int("abc") would crash your program with a ValueError, because "abc" is not a valid number.
# We are not running it here on purpose - feel free to try it yourself in a new cell!
````

### Example 6: Making Decisions with if / elif / else

An `if` statement lets your program take different paths depending on a condition. Python checks the conditions from top to bottom and runs the code below the **first** one that is `True`. `elif` means "else if", and `else` catches everything else. **Indentation matters in Python**: the indented lines below `if`/`elif`/`else` are what belongs to that branch — this is how Python knows where a block of code starts and ends (instead of using curly braces `{}` like many other languages).

````python
temperature = 28

if temperature >= 30:
    print("It is hot outside.")
elif temperature >= 20:
    print("It is warm outside.")   # This branch runs, since 20 <= 28 < 30
elif temperature >= 10:
    print("It is cool outside.")
else:
    print("It is cold outside.")
````

### Example 7: Repeating Code with Loops

A `for` loop repeats a block of code once for each element in a sequence (like a list, or the numbers produced by `range()`). This is extremely common when working with text data, e.g. "do something for every word in this sentence".

````python
fruits = ["apple", "banana", "mango"]

# for loop over a list -> "fruit" takes on each value in the list, one at a time
for fruit in fruits:
    print(f"I like {fruit}s")

print()

# range(5) produces the numbers 0, 1, 2, 3, 4 (5 is NOT included)
for i in range(5):
    print(f"Count: {i}")
````

A `while` loop repeats a block of code for as long as its condition stays `True`. It's useful when you don't know in advance how many times you need to repeat something. `break` stops the loop immediately, and `continue` skips straight to the next iteration.

````python
countdown = 5

while countdown > 0:
    if countdown == 3:
        print("(skipping 3 with continue)")
        countdown -= 1  # same as: countdown = countdown - 1
        continue        # jumps back to the "while" condition, skipping the print() below
    print(countdown)
    countdown -= 1

print("Liftoff!")
````

### Example 8: String Operations and F-Strings

Strings have many built-in methods for cleaning and transforming text — very useful for NLP tasks, which we'll get to in the next notebook.

````python
# String operations
text = "  Python for Natural Language Processing  "

print(f"Original: '{text}'")
print(f"Length: {len(text)}")
print(f"Uppercase: {text.upper()}")
print(f"Lowercase: {text.lower()}")
print(f"Without whitespace: '{text.strip()}'")  # strip() removes leading/trailing spaces
print(f"Replaced: {text.replace('Python', 'Java')}")

# Checking the content of a string
print(f"Contains 'Language': {'Language' in text}")     # the "in" keyword checks for a substring
print(f"Starts with '  Py': {text.startswith('  Py')}")

# Split string into a list of words (default separator is whitespace)
words = text.strip().split()
print(f"Words: {words}")
````

````python
# F-String formatting: {value:spec} lets you control how a value is displayed
name = "Maria"
points = 95.7

print(f"Student {name} has achieved {points:.1f} points.")     # .1f -> 1 digit after the decimal point
print(f"That is {points/100:.1%} of the possible points.")     # .1% -> shown as a percentage
````

### Example 9: List Operations

A `list` is an ordered, changeable collection of values, written with square brackets `[]`. Elements are accessed by their position (**index**), starting at `0` for the first element.

````python
numbers = [1, 3, 5, 7, 9]
print(f"Full list: {numbers}")
print(f"First element (index 0): {numbers[0]}")
print(f"Last element (index -1): {numbers[-1]}")  # negative indices count from the end

# Slicing: list[start:stop] -> "stop" is NOT included
print(f"First 3 elements: {numbers[:3]}")
print(f"Last 3 elements: {numbers[-3:]}")
print(f"Every second element: {numbers[::2]}")
````

````python
# Adding elements
numbers.append(11)          # append() adds a single element at the end
numbers.extend([13, 15])    # extend() adds all elements of another list
numbers.insert(0, -1)       # insert(position, value) adds at a specific position
print(f"After adding: {numbers}")

# Removing elements
numbers.remove(-1)          # remove() deletes the first occurrence of a specific value
last_element = numbers.pop() # pop() removes AND returns the last element (or any index you give it)
print(f"After removing: {numbers}")
print(f"Removed with pop(): {last_element}")
````

````python
# List comprehension: a compact way to build a new list from an existing one
squares = [x**2 for x in numbers if x > 0]  # squares of the positive numbers in the list
print(f"Squares of positive numbers: {squares}")

# Sorting
numbers.sort(reverse=True)  # sort in descending order, in place
print(f"Sorted (descending): {numbers}")
````

### Example 10: Tuples and Sets

A `tuple` looks like a list but uses round brackets `()` and, once created, **cannot be changed** (it's "immutable"). It's often used for values that belong together and shouldn't accidentally be modified, like a coordinate pair.

A `set` is an unordered collection that automatically removes duplicates — handy for tasks like "give me all the unique words in this text".

````python
# Tuples: fixed, ordered, written with ()
coordinates = (52.5200, 13.4050)  # e.g. latitude and longitude of Berlin
latitude, longitude = coordinates  # "unpacking" a tuple into separate variables
print(f"Latitude: {latitude}, Longitude: {longitude}")

# coordinates[0] = 0  # this would raise an error - tuples cannot be modified after creation
````

````python
# Sets: unordered, written with {}, automatically remove duplicates
words = ["python", "is", "fun", "python", "is", "great"]
unique_words = set(words)
print(f"Original list ({len(words)} words): {words}")
print(f"Unique words ({len(unique_words)} words): {unique_words}")

# Sets also support typical set operations
beginner_topics = {"variables", "loops", "functions"}
advanced_topics = {"functions", "classes", "decorators"}
print(f"In both: {beginner_topics & advanced_topics}")   # intersection
print(f"In either: {beginner_topics | advanced_topics}") # union
````

### Example 11: Dictionaries for Text Processing

A `dict` (dictionary) stores **key-value pairs**, written with curly brackets `{}`. Instead of accessing values by position like in a list, you access them by their key — which makes dictionaries perfect for things like counting how often each word appears in a text.

````python
address = {"city": "Munich", "zip": 80331}

print(f"City: {address['city']}")   # access a value by its key
print(f"Keys: {list(address.keys())}")
print(f"Values: {list(address.values())}")
print(f"Items: {list(address.items())}")

# .get() is a safe way to access a key that might not exist - it returns a default instead of crashing
print(f"Country: {address.get('country', 'Unknown')}")
````

````python
# Now it gets a bit more complex: we will count the frequency of words in a text.
text = "Python is a great language. Python is fun. Programming with Python is easy."

# Split text into words and clean it
words = text.lower()               # convert the text to lowercase
words = words.replace('.', '')     # remove punctuation "."
words = words.replace(',', '')     # remove punctuation ","

# Now let's convert the text (string) into a list of words. The default separator is whitespace.
words = words.split()
print(f"Cleaned words: {words}")
````

````python
# Count word frequencies by iterating over the list of words
word_frequency = {}
for word in words:
    if word in word_frequency:      # check if the word is already in the dictionary
        word_frequency[word] += 1   # if so, increment its counter by 1
    else:
        word_frequency[word] = 1    # if not, initialize its counter with 1

print(f"Word frequencies: {word_frequency}")

# The same loop written more compactly using dict.get(word, 0), which returns 0 if the word is new
word_frequency_v2 = {}
for word in words:
    word_frequency_v2[word] = word_frequency_v2.get(word, 0) + 1

print(f"Same result: {word_frequency_v2}")
````

### Example 12: Functions

A function is a reusable, named block of code, defined with `def`. Functions can take **parameters** (inputs), optionally with **default values**, and give back a result with `return`. Wrapping code in functions avoids repeating yourself and makes programs easier to read.

````python
def convert_celsius_to_fahrenheit(celsius, round_result=True):
    """Converts a Celsius temperature to Fahrenheit.
    "round_result" has a default value, so it is optional when calling the function."""
    fahrenheit = celsius * 9 / 5 + 32
    if round_result:
        fahrenheit = round(fahrenheit, 1)
    return fahrenheit

print(convert_celsius_to_fahrenheit(20))                 # uses the default round_result=True
print(convert_celsius_to_fahrenheit(20, round_result=False))
````

````python
def summarize_numbers(*numbers):
    """The * in *numbers lets this function accept any number of arguments,
    which Python collects into a tuple called "numbers"."""
    return f"Sum: {sum(numbers)}, Average: {sum(numbers) / len(numbers):.2f}"

print(summarize_numbers(4, 8, 15))
print(summarize_numbers(1, 2, 3, 4, 5, 6))
````

### Example 13: Classes and Objects (Introduction to OOP)

A `class` is a blueprint for creating objects that bundle together data (**attributes**) and behavior (**methods**). `__init__` is a special method (the **constructor**) that runs automatically when a new object is created, and `self` refers to "this particular object".

````python
class Animal:
    def __init__(self, name, sound):
        """Constructor: stores the attributes on the object (self)"""
        self.name = name
        self.sound = sound

    def make_sound(self):
        """A method: a function that belongs to the object and can use its attributes via self"""
        return f"{self.name} says {self.sound}!"

generic_animal = Animal("Some animal", "...")
print(generic_animal.make_sound())
````

**Inheritance** lets a class reuse and extend another class, using `class Dog(Animal):`. Here, `Dog` inherits everything `Animal` already has, and can override a method with its own version — this is called **polymorphism**: the same method name (`make_sound`) behaves differently depending on the object's actual class.

````python
class Dog(Animal):
    def __init__(self, name):
        super().__init__(name, sound="Woof")  # reuse the parent constructor (Animal.__init__)

    def make_sound(self):
        """Overriding the parent method (Polymorphism)"""
        return f"{self.name} the dog barks: {self.sound}!"

my_dog = Dog("Rex")
print(my_dog.make_sound())        # uses Dog's own version

# Polymorphism in action: the same method call, different behavior per object
animals = [generic_animal, my_dog]
for animal in animals:
    print(animal.make_sound())
````

```python
# 1. Average
average = sum(student_grades.values()) / len(student_grades)

# 2. Best & worst
best = min(student_grades, key=student_grades.get)
worst = max(student_grades, key=student_grades.get)

# 3. Students with grade < 2.0
best_students = [s for s, g in student_grades.items() if g < 2.0]

# 4. New dictionary with ratings
def rating(grade):
    if grade <= 1.5: return "Excellent"
    elif grade <= 2.5: return "Good"

for s, g in student_grades.items():
    ratings[s] = rating(g)

print(f"Best: {best}, Worst: {worst}")
print("Best students:", best_students)
print("Ratings:", ratings)
```

```python
# Base class Person
class Person:
    def __init__(self, name, age):
        """Constructor: automatically called when creating an object"""
        self.name = name
        self.age = age
    
    def greet(self):
        """Greeting method"""
        return f"Hello, my name is {self.name} and I am {self.age} years old."
    
    def is_of_legal_age(self):
        """Checks if person is of legal age"""
        return self.age >= 18

# Inheritance: Student inherits from Person
class Student(Person):
    def __init__(self, name, age, major):
        """Constructor with super() to call parent constructor"""
        super().__init__(name, age)  # Call parent constructor
        self.major = major
    
    def greet(self):
        """Overridden greet method (Polymorphism)"""
        return f"Hello, my name is {self.name}, I am {self.age} years old and I study {self.major}."
    
    def study_info(self):
        """Additional method only for students"""
        return f"{self.name} studies {self.major}."

# Tests of the classes
print("=== Test Person Class ===")
person1 = Person("Max Mustermann", 25)
print(person1.greet())
print(f"Of legal age: {person1.is_of_legal_age()}")

person2 = Person("Lisa Schmidt", 16)
print(person2.greet())
print(f"Of legal age: {person2.is_of_legal_age()}")

print("\n=== Test Student Class (Inheritance) ===")
student1 = Student("Anna Miller", 22, "Computer Science")
print(student1.greet())  # Overridden method
print(f"Of legal age: {student1.is_of_legal_age()}")  # Inherited method
print(student1.study_info())  # Own method

student2 = Student("Tom Weber", 17, "Mathematics")
print(student2.greet())
print(f"Of legal age: {student2.is_of_legal_age()}")

print("\n=== Polymorphism Demonstration ===")
persons = [person1, student1, student2]
for p in persons:
    print(p.greet())  # Each class has its own greet() implementation
```
