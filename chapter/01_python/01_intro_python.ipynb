### 7. Variables Are Names, Not Copies

`b = a` does **not** copy the list. `a` and `b` are two names for the same list.

````python
a = [1, 2, 3]
b = a
b.append(4)
print(a)  # a has changed too!
````

`copy()` creates a real, independent copy.

````python
c = a.copy()
c.append(5)
print(a)
print(c)
````

`==` checks if two values are **equal**. `is` checks if they are **the same** object.

````python
x = [1, 2, 3]
y = [1, 2, 3]
print(x == y)  # True: same values
print(x is y)  # False: two different lists
````

Rule of thumb: use `is` only for `None`, otherwise use `==`.

````python
result = None
print(result is None)
````

### 12. Loops

A **while loop** repeats code as long as the condition is `True`.

````python
count = 1
while count <= 3:
    print(count)
    count += 1  # same as: count = count + 1
````

A **for loop** goes through each element of a list.

````python
for fruit in ["apple", "banana", "mango"]:
    print(fruit)
````

It also works with a string (character by character)...

````python
for letter in "abc":
    print(letter)
````

...and with a dictionary. `.items()` gives you key and value.

````python
grades = {"Anna": 1.3, "Ben": 2.7}
for name, grade in grades.items():
    print(name, grade)
````

`break` stops the loop.

````python
for number in [1, 2, 3, 4, 5]:
    if number == 3:
        break
    print(number)
````

`continue` skips to the next round.

````python
for number in [1, 2, 3, 4, 5]:
    if number == 3:
        continue
    print(number)
````

`range(n)` gives the numbers `0` to `n-1`.

````python
for i in range(5):
    print(i)
````

`range(start, stop)` starts at a different number.

````python
for i in range(2, 5):
    print(i)
````

#### List Comprehensions

A **list comprehension** builds a new list in one line.

````python
numbers = [1, 2, 3, 4]
squares = [x * x for x in numbers]
print(squares)
````

With `if` you can filter elements.

````python
even = [x for x in numbers if x % 2 == 0]
print(even)
````

### 14. Functions

A **function** is a reusable block of code. You define it with `def`.

````python
def say_hello():
    print("Hello!")
````

Call the function with its name and `()`.

````python
say_hello()
````

A function can take **arguments** and give back a result with `return`.

````python
def add(a, b):
    return a + b

print(add(3, 4))
````

A parameter can have a **default value**. Then it is optional.

````python
def greet(name, greeting="Hello"):
    print(f"{greeting}, {name}!")

greet("Anna")
greet("Anna", "Hi")
````

With **keyword arguments** you pass values by name. The order does not matter.

````python
def introduce(name, age):
    print(f"{name} is {age} years old.")

introduce(age=30, name="Anna")
````

Positional arguments must come **before** keyword arguments.

````python
def info(name, city, language):
    print(f"{name} lives in {city} and speaks {language}.")

info("Lina", "Berlin", language="German")       # ok
# info(name="Lina", "Berlin", language="German")  -> SyntaxError
````

With `*args` a function accepts **any number** of arguments.

````python
def total(*numbers):
    return sum(numbers)

print(total(1, 2))
print(total(1, 2, 3, 4))
````

### 15. Scopes

### 16. Classes and Objects

A **class** is a blueprint for objects. `__init__` runs when an object is created.

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
````

Create an **object** and access its **attributes**.

````python
anna = Person("Anna", 29)
print(anna.name)
print(anna.age)
````

A **method** is a function inside a class. `self` is the object itself.

````python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        return f"Hi, I am {self.name}."
````

````python
anna = Person("Anna")
print(anna.greet())
````

**Inheritance**: `Student` gets everything from `Person`. `super().__init__()` calls the parent constructor.

````python
class Student(Person):
    def __init__(self, name, major):
        super().__init__(name)
        self.major = major
````

````python
ben = Student("Ben", "Computer Science")
print(ben.greet())  # inherited from Person
print(ben.major)
````

**Polymorphism**: a child class can override a method. The same call then behaves differently.

````python
class Student(Person):
    def greet(self):
        return f"Hi, I am {self.name} and I am a student."
````

````python
for p in [Person("Anna"), Student("Ben")]:
    print(p.greet())
````
