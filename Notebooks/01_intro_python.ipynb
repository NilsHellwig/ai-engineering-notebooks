### Example 1: Variables and Data Types

````python
# Different data types in Python
name = "Max"  # String
age = 25      # Integer
height = 1.68  # Float
is_student = True  # Boolean
hobbies = ["Reading", "Programming", "Sports"]  # List
address = {"city": "Munich", "zip": 80331}  # Dictionary

# Now let's use print() to output the variables and their types. Reminder: f-Strings (f"<...>") allow us to insert variables directly into the string.
print(f"Name: {name} (Type: {type(name)})")
print(f"Age: {age} (Type: {type(age)})")
print(f"Height: {height} (Type: {type(height)})")
print(f"Is Student: {is_student} (Type: {type(is_student)})")
print(f"Hobbies: {hobbies} (Type: {type(hobbies)})")
print(f"Address: {address} (Type: {type(address)})")
````

### Example 2: String Operations and F-Strings

````python
# String operations
# Now we want to demonstrate some basic string operations.

text = "  Python for Natural Language Processing  "

print(f"Original: '{text}'")
print(f"Length: {len(text)}")
print(f"Uppercase: {text.upper()}")
print(f"Lowercase: {text.lower()}")
print(f"Without whitespace: '{text.strip()}'")
print(f"Replaced: {text.replace('Python', 'Java')}")

# Split string
words = text.strip().split()
print(f"Words: {words}")

# F-String formatting
name = "Maria"
points = 95.7
print(f"Student {name} has achieved {points:.1f} points.")
print(f"That is {points/100:.1%} of the possible points.")
````

### Example 3: List Operations

````python
# Create and manipulate lists
numbers = [1, 3, 5, 7, 9]
print(f"Original list: {numbers}")

# Add elements
numbers.append(11) # append() adds a single element
numbers.extend([13, 15]) # extend() adds a list
numbers.insert(0, -1) # insert() adds at a specific position
print(f"After adding: {numbers}")

# Slicing
print(f"First 3 elements: {numbers[:3]}") # :3 selects the first three elements
print(f"Last 3 elements: {numbers[-3:]}") # -3: selects the last three elements
print(f"Every second element: {numbers[::2]}") # ::2 selects every second element

# List Comprehension
squares = [x**2 for x in numbers if x > 0] # Squares of positive numbers in the list
print(f"Squares of positive numbers: {squares}")

# Sorting
numbers.sort(reverse=True) # Sort in descending order
print(f"Sorted (descending): {numbers}")
````

### Example 4: Dictionaries for Text Processing

````python
# Dictionary for word frequencies
# Now it gets a bit more complex: We will count the frequency of words in a text.

text = "Python is a great language. Python is fun. Programming with Python is easy."

# Split text into words and clean it
words = text.lower() # Convert the text to lowercase
words = words.replace('.', '') # Remove punctuation "."
words = words.replace(',', '') # Remove punctuation ","

# Now let's convert the text (string) into a list of words. The default separator is whitespace. 
# If we pass an argument to the .split() method, it will split at that character.
# If we used .split(','), for example, the text would be split at each comma.
words = words.split()
print(f"Cleaned words: {words}")

# Count word frequencies. Let's iterate over the list of words and use a dictionary to store the frequencies.
word_frequency = {}
for word in words:
    if word in word_frequency: # Check if the word is already in the dictionary
        # If the word is already in the dictionary, increment the counter by 1. 
        word_frequency[word] += 1
    else:
        # If the word is not yet in the dictionary, initialize the counter with 1
        word_frequency[word] = 1

print(f"Word frequencies: {word_frequency}")
````

average = sum(temperatures_celsius) / len(temperatures_celsius)
print(f"Average temperature: {average:.1f}°C")

# 3. Temperatures above 25°C
warm_days = [temp for temp in temperatures_celsius if temp > 25]

### Exercise 3: Dictionary Operations

**Task:** You have a dictionary with students and their grades. Write a program that:
1. Calculates the average grade of all students
2. Finds the best and worst student
3. Outputs all students with a grade better than 2.0
4. Creates a new dictionary with ratings (1.0-1.5: "Excellent", 1.6-2.5: "Good", etc.)

````python
# Given students and grades
student_grades = {
    "Anna": 1.3,
    "Ben": 2.7,
    "Clara": 1.0,
    "David": 3.2,
    "Eva": 1.8,
    "Felix": 2.1
}

# Your solution here:
````

<details>
<summary><b>Show Solution</b></summary>

```python
# 1. Average
average = sum(student_grades.values()) / len(student_grades)

# 2. Best & worst
best = min(student_grades, key=student_grades.get)
worst = max(student_grades, key=student_grades.get)

# 3. Students with grade < 2.0
best_students = [s for s, g in student_grades.items() if g < 2.0]

# 4. New dictionary with ratings
def rating(grade):
    if grade <= 1.5: return "Excellent"
    elif grade <= 2.5: return "Good"
    elif grade <= 3.5: return "Satisfactory"
    else: return "Sufficient"

ratings = {}
for s, g in student_grades.items():
    ratings[s] = rating(g)

print(f"Average: {average:.2f}")
print(f"Best: {best}, Worst: {worst}")
print("Best students:", best_students)
print("Ratings:", ratings)
```

The :.2f in an f-String (e.g., f"{average:.2f}") is a formatting specifier and means:

: → introduces the format specification
.2 → round to 2 decimal places
f → format number as floating-point. Alternatively, there would also be e.g., e for exponential notation.

### Exercise 5: Classes and Objects

**Task:** Create a class `Person` based on the concepts from the slides:

1. The class should have a constructor that initializes the attributes `name` and `age`.
2. A method `greet` that returns a greeting.
3. A method `is_of_legal_age(self)` that checks if the person is >= 18 years old.

**Advanced tasks:**

4. Then create a class `Student` that inherits from Person.
5. The Student class should additionally have an attribute `major`.
6. Override the `greet()` method to also output the major.
7. Test both classes with different objects.

````python
# Your solution here:
````

<details>
<summary><b>Show Solution</b></summary>

```python
# Base class Person
class Person:
    def __init__(self, name, age):
        """Constructor: automatically called when creating an object"""
        self.name = name
        self.age = age
    
    def greet(self):
        """Greeting method"""
        return f"Hello, my name is {self.name} and I am {self.age} years old."
    
    def is_of_legal_age(self):
        """Checks if person is of legal age"""
        return self.age >= 18

# Inheritance: Student inherits from Person
class Student(Person):
    def __init__(self, name, age, major):
        """Constructor with super() to call parent constructor"""
        super().__init__(name, age)  # Call parent constructor
        self.major = major
    
    def greet(self):
        """Overridden greet method (Polymorphism)"""
        return f"Hello, my name is {self.name}, I am {self.age} years old and I study {self.major}."
    
    def study_info(self):
        """Additional method only for students"""
        return f"{self.name} studies {self.major}."

# Tests of the classes
print("=== Test Person Class ===")
person1 = Person("Max Mustermann", 25)
print(person1.greet())
print(f"Of legal age: {person1.is_of_legal_age()}")

person2 = Person("Lisa Schmidt", 16)
print(person2.greet())
print(f"Of legal age: {person2.is_of_legal_age()}")

print("\n=== Test Student Class (Inheritance) ===")
student1 = Student("Anna Miller", 22, "Computer Science")
print(student1.greet())  # Overridden method
print(f"Of legal age: {student1.is_of_legal_age()}")  # Inherited method
print(student1.study_info())  # Own method

student2 = Student("Tom Weber", 17, "Mathematics")
print(student2.greet())
print(f"Of legal age: {student2.is_of_legal_age()}")

print("\n=== Polymorphism Demonstration ===")
persons = [person1, student1, student2]
for p in persons:
    print(p.greet())  # Each class has its own greet() implementation
```
